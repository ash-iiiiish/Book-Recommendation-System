```python
import pandas as pd

df = pd.read_csv("C:/Users/kumar/OneDrive/Desktop/TRY-2/Tasks-Submission/Assignment20/data.csv")

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 rows:")
print(df.head())
print("\nDataset Info:")
print(df.info())
```

# Task2

```python
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    tokens = [word for word in text.split() if word not in ENGLISH_STOP_WORDS]
    return " ".join(tokens)

df["text"] = (
    df["title"].fillna("") + " " +
    df["authors"].fillna("") + " " +
    df["categories"].fillna("") + " " +
    df["description"].fillna("")
)

df["clean_text"] = df["text"].apply(clean_text)
```

# Part2

# Task3

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)

tfidf_matrix = tfidf.fit_transform(df["clean_text"])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

# Task4

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix)
```

# Part3

# Task5

```python
def recommend(item_name, top_n=5):
    if item_name not in df["title"].values:
        return ["Book not found"]

    idx = df[df["title"] == item_name].index[0]
    similarity_scores = list(enumerate(cosine_sim[idx]))
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)

    top_books = similarity_scores[1:top_n+1]
    recommendations = [df.iloc[i[0]]["title"] for i in top_books]

    return recommendations
```

```python
print(recommend("The Hobbit"))
print(recommend("Harry Potter and the Sorcerer's Stone"))
print(recommend("The Alchemist"))
```
